# Model Card: Arabic-to-English Emotion Classification Transformer

## 1. Model Overview

This is a fine-tuned transformer-based classifier based on the CAMeL-Lab Arabic BERT model (bert-base-arabic-camelbert-da-sentiment) All the sentences that were originally spoken in Arabic and then translated into English, are tagged in this database with the emotional label they refer. The model predicts one of the emotions below for each input: happiness, sadness, anger, surprise, fear, disgust, or neutral.


## 2. Architecture

- **Base Model**: CAMeL-Lab's BERT model for Arabic sentiment
- **Task Type**: Single-label sequence classification
- **Tokenizer**: Hugging Face tokenizer with max sequence length of 128
- **Loss Function**: CrossEntropyLoss
- **Optimizer**: AdamW (default via Hugging Face Trainer)
- **Metrics**: Macro F1 score and validation loss
- **Training Epochs**: 6

## 3. Purpose

The model was created for the Content Intelligence Agency, which wants to understand media content like TV shows and movies. A model that focuses on the phase of emotion recognition from the video transcript translation. With this model, we aim to create an automated way of annotating content in terms of emotional registers in order to also assist the analysis of how we as humans may feel in relation to certain scenes or dialogues.

## 4. Development Context

Training and evaluation of the model occurred over about 3 weeks. Initially, our performance was lower because of mapping issues regarding emotion labels and limited access to quality emotion-labeled Arabic datasets. A related challenge was the wide variety of Arabic dialects, which meant that translations were not always straightforward. Translation quality also affected the degree to which the English text preserved the original emotional tone. The work was carried out in a GPU setting, using Python libraries including Hugging Face Transformers and PyTorch.

## 5. Intended Use

The output data is designed to be used on translated video transcripts so as to recognize sentimental lines in a video. It can also be used to annotate emotions in media content and assist media analysts in understanding how an emotional tone shifts over the course of a show, scene, etc. Such feature can be useful for storytelling, content curation or audience response analysis.

It should not be used in applications such as mental health diagnosis or where misclassification could have real-world effects. The model does not have real-time emotion tracking capability, and is not ready for direct deployment without a human reviewer. Best used behind-the-scenes in a more complex, broader content analysis work flow.

This model supports the client’s media analysis goals by enabling the Content Intelligence Agency to scale up their emotional annotation process. It helps reduce manual labor and offers a way to analyze emotional narratives across translated content, including from Arabic-speaking regions.

## 6. Dataset Details

Two datasets of translated sentences and their associated emotion labels were utilized for training and testing the model. We trained this model first on the first video datset that we recived from the agency and then predicted on the second video received from second video. The first sentence was translated from Arabic to English and assigned one of seven emotion labels.

-Selected the non-null rows and mapped the textual labels to integers.

## 7. Performance Metrics and Evaluation

The model achieved the following results:

- **Best F1 Score**: 0.768 (epoch 3)
- **Final F1 Score**: 0.7449 (epoch 6)
- **Validation Loss**: 0.6647 (epoch 6)

The confusion matrix shows a strong bias toward predicting the "neutral" class. Emotions such as fear, disgust, and sadness were often misclassified or confused with neutral. This suggests the model struggles more with less frequent or subtle emotions.

## 8. Explainability and Transparency

In Task 9, Gradient × Input was used as a technique to understand model decisions. This involved analyzing the gradient of the predicted class with respect to each token’s embedding and then multiplying the gradients by the input. This method allowed us to visualize which words had the most influence on the final classification. This allowed us to visualize which words or expressions had the most influence on the final classification and how the model internally associates certain words or expressions with  emotional categories.

## 9. Recommendations for Use

- The model is most effective when used as part of a human-in-the-loop system.
- It should not be fully relied upon for high-stakes decision-making.
- Media analysts should consider reviewing model predictions, especially when dealing with emotions that are nuanced or underrepresented in the training data.
- Better results may be obtained by improving translation consistency or using back-translation to check emotional fidelity.

## 10. Sustainability Considerations

- **Training Time**: ~1 hour
- **Hardware**: Trained on a GPU
- **Carbon Footprint**: Not measured during this project

While the training session was relatively short, the environmental impact could be reduced by:
- Using cloud resources in regions powered by renewable energy
- Applying early stopping or dynamic learning rate scheduling to minimize training epochs
- Reusing trained models for similar tasks to avoid redundant training
