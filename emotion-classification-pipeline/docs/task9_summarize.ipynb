# Model Interpretability Summary for Emotion Classification

## Introduction

In this analysis, we aim to interpret the decision-making process of a multilingual emotion classification model. Emotion classification, which involves detecting emotions from text, is a complex task due to the subjective nature of emotions and the nuances of language. We explore how the model assigns emotional relevance to different tokens in sentences using two explanation techniques: Gradient × Input and Layer-wise Relevance Propagation (LRP). These methods help identify which tokens (words or phrases) influence the model’s emotional predictions. The investigation focuses on how well these interpretations align with human expectations of emotional content. This summary is meant to be included and linked in the model card for transparency and interpretability reporting. By doing so, we ensure that the model's decision-making process is not only effective but also understandable to its users.

## Analysis of Token Relevance

We begin with a happiness-labeled sentence in Arabic: "الفرح كبسها" (translated as "Joy pressed her"). In the visualization titled "Token Relevance for 'الفرح كبسها' (Predicted Emotion: happiness)", we observe how the model interprets each token. The token "كبسها" (pressed her) had the highest positive influence, while "الفرح" (joy) surprisingly showed strong negative relevance, despite being part of the emotionally significant action. Special tokens like [CLS] and [SEP] displayed minimal relevance, which aligns with expectations since they typically serve formatting roles rather than semantic ones. However, this indicates a partial misalignment with human understanding, as "press" should arguably contribute to the emotional label of happiness. The presence of such unexpected relevance patterns emphasizes the need for further refinement of the model’s ability to interpret emotional contexts accurately.

In the English translation of the same sentence, represented in the figure titled "Token Relevance for 'Joy pressed her' (Predicted Emotion: happiness)", the model assigned most of the positive relevance to the token "her," while the action word "pressed" had lower-than-expected influence. From a human perspective, this is unexpected, as the verb in the sentence often carries the emotional tone. This reveals that the model may not be placing appropriate weight on action-oriented tokens, which could impact its performance in emotionally charged contexts. For example, in many emotional statements, action words (verbs) often hold significant weight in conveying the emotional tone. Therefore, the model’s lower emphasis on the verb "pressed" suggests that the model's interpretability methods may need to be refined further to better capture action-driven emotion signals.

We further investigated model behavior through perturbation analysis. The figure "Model Confidence with Remove Perturbation" tracks how model confidence changes when tokens are systematically removed from the input. Initially, as individual tokens are removed, the model's confidence fluctuates, with a sharp drop around the fourth token. Interestingly, after removing the fifth token, the model’s confidence spikes and then stabilizes. This suggests that the model heavily relies on a small set of tokens and may even increase confidence when potentially distracting or noise-inducing words are removed. However, the stability in confidence after the removal of multiple tokens could also indicate brittle or oversimplified reasoning processes. The model’s reliance on specific tokens, rather than considering the full sentence context, could result in lower generalization capacity when faced with more complex emotional inputs.

The surprise emotion example further highlighted inconsistencies. We analyzed the sentence "فجأة صارت أربعة" (translated: "Suddenly it became four"). This sentence intuitively carries a tone of surprise, primarily due to the word "فجأة" (suddenly). However, the model predicted a neutral emotion. In the figure titled "Gradient × Input and LRP for 'فجأة صارت أربعة' (Emotion: neutral)", the Gradient × Input method assigned negative relevance to "فجأة," while LRP assigned high relevance to [SEP], a special token with no semantic contribution. These findings indicate a significant misalignment between human emotion perception and model predictions. Despite the obvious trigger word for surprise, "فجأة," the model appeared to downplay this emotional cue, resulting in a neutral classification. This discrepancy highlights a key challenge in emotion classification, where models may not always capture the subtle emotional cues that humans intuitively identify.

Overall, these examples suggest that the model sometimes fails to properly weigh emotionally significant tokens, especially across different languages. Words that are semantically critical to expressing an emotion—like action verbs or modifiers—are occasionally downplayed or even misinterpreted. Additionally, model explanations sometimes place undue emphasis on structurally insignificant tokens like [SEP], pointing to training artifacts or model overfitting. Such issues are particularly important for multilingual models, where the same emotional concepts may be expressed in different ways across languages, requiring the model to be adaptable and sensitive to linguistic variations.

Furthermore, confidence analysis shows that the model’s predictions are not always robust to minor changes. The confidence trend following token removal revealed sensitivity and inconsistency, especially early in the sequence. While stability is reached after a few tokens, it might indicate a shortcut or oversimplification in decision-making. This behavior suggests that the model may be overly confident in certain situations, but that confidence may not always be justified, especially when it comes to more complex or subtle emotional nuances.

## Conclusion

This interpretability study reveals the inner workings and limitations of a multilingual emotion classification model. While some outputs align with human expectations, there are several clear mismatches, particularly in cases of surprise and happiness emotions. These findings underscore the importance of rigorous interpretability testing, especially for models deployed in real-world, emotionally sensitive applications. With these insights, future development can focus on improving the model’s alignment with human semantic and emotional understanding, thus enhancing its reliability and trustworthiness. Ongoing improvements in interpretability techniques and fine-tuning approaches will be crucial for the continued advancement of emotion-aware systems.
