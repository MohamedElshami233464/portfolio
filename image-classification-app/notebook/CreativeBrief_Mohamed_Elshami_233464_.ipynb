# Grad-CAM

````python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import cv2
from keras.models import load_model


# Identify a convolutional layer in your model
layer_name = 'conv2d_1'  

# Function to compute Grad-CAM
def compute_gradcam(model, img, layer_name):
    grad_model = tf.keras.models.Model([model.inputs], [model.get_layer(layer_name).output, model.output])
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(np.array([img]))
        loss = predictions[:, np.argmax(predictions[0])]

    output = conv_outputs[0]
    grads = tape.gradient(loss, conv_outputs)[0]
    gate_f = tf.cast(output > 0, 'float32')
    gate_r = tf.cast(grads > 0, 'float32')
    guided_grads = gate_f * gate_r * grads

    weights = tf.reduce_mean(guided_grads, axis=(0, 1))
    cam = np.ones(output.shape[0:2], dtype=np.float32)

    for i, w in enumerate(weights):
        cam += w * output[:, :, i]

    cam = cv2.resize(cam.numpy(), (img.shape[1], img.shape[0]))
    cam = np.maximum(cam, 0)
    heatmap = (cam - cam.min()) / (cam.max() - cam.min())

    return heatmap

# Select an image to explain
image_to_explain = X_test_v1_rgb[3]

# Compute Grad-CAM
heatmap = compute_gradcam(model_v1, image_to_explain, layer_name)

# Display the Grad-CAM heatmap
plt.imshow(image_to_explain)
plt.imshow(heatmap, cmap='jet', alpha=0.5)  # overlay heatmap
plt.title('Grad-CAM Explanation')
plt.axis('off')
plt.show()
````


### Explainable AI Methods Used:
1. LIME (Local Interpretable Model-agnostic Explanations)
2. Grad-CAM (Gradient-weighted Class Activation Mapping)

### Applying LIME
LIME helps to explain individual predictions by approximating the model locally with an interpretable model. Here, LIME is applied to visualize the regions in the image that the model considers important for its decision.


### Applying Grad-CAM
Grad-CAM provides visual explanations for convolutional neural networks by highlighting important regions in the input image. This helps in understanding which parts of the image are most influential in the model’s decision.


### Trade-off Between Accuracy and Interpretability
While LIME and Grad-CAM enhance the interpretability of the model by providing visual explanations, they come with trade-offs:
- **LIME**: Provides detailed local explanations but can be computationally intensive.
- **Grad-CAM**: Offers a global view of feature importance in CNNs but may not be as precise in highlighting specific regions.

### Iteration 1

For error analysis, I identified and plotted wrong classifications for the initial model.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Function to identify and plot wrong classifications
def plot_wrong_classifications(model_v1, X_test_v1_rgb, y_test_v1, class_labels):
    # Get predictions on the test set
    predictions = model_v1.predict(X_test_v1_rgb)
    predicted_labels = np.argmax(predictions, axis=1)
    actual_labels = np.argmax(y_test_v1, axis=1)

    # Find instances where predictions do not match labels
    mismatches = np.where(predicted_labels != actual_labels)[0]
    wrong_predictions = X_test_v1_rgb[mismatches]
    wrong_actual_labels = actual_labels[mismatches]
    wrong_predicted_labels = predicted_labels[mismatches]

    print("Wrong actual labels:", wrong_actual_labels)
    print("Wrong predicted labels:", wrong_predicted_labels)

    # Plot wrong classifications
    plt.figure(figsize=(12, 8))
    for i in range(len(wrong_predictions)):
        plt.subplot(3, 5, i + 1)
        plt.imshow(wrong_predictions[i])
        plt.title(f"Actual: {class_labels[wrong_actual_labels[i]]}\nPredicted: {class_labels[wrong_predicted_labels[i]]}")
        plt.axis('off')
    plt.tight_layout()
    plt.show()

# Function to categorize errors based on common characteristics
def categorize_errors(X_test_v1_rgb, y_test_v1, model_v1):
    # Get predictions on the test set
    predictions = model_v1.predict(X_test_v1_rgb)
    predicted_labels = np.argmax(predictions, axis=1)
    actual_labels = np.argmax(y_test_v1, axis=1)

    # Compute confusion matrix
    cm = confusion_matrix(actual_labels, predicted_labels)

    # Plot confusion matrix
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, cmap='Blues', fmt='g', xticklabels=class_labels, yticklabels=class_labels)
    plt.xlabel('Predicted Labels')
    plt.ylabel('Actual Labels')
    plt.title('Confusion Matrix')
    plt.show()

# Define class labels
class_labels = ['Fresh banana', 'Rotten banana', 'Fresh apple', 'Rotten apple']

# Example usage
plot_wrong_classifications(model_v1, X_test_v1_rgb, y_test_v1, class_labels)
categorize_errors(X_test_v1_rgb, y_test_v1, model_v1)
````
